Distributing Candies to Children

```python
def candy(ratings):
    n=len(ratings)
    cand = [1]*n

    #left to right
    for i in range(1 , n):
        if ratings[i] > ratings[i - 1]:
            cand[i] = cand[i - 1]+1
    #right to left
    for i in range(n-2, -1 ,-1):
        if ratings[i] > ratings[i+1]:
            cand[i] = max(cand[i],cand[i+1]+1)
    
    return sum(cand)

#input
size = int(input())
ratings = list(map(int,input().split()))
print(candy(ratings))
```

BUY AND SELL STOCKS 

```python
def max_profit(arr):
    if not arr:
        return 0
    min = float('inf') # set very high number
    max = 0
    for i in range(len(arr)):
        if arr[i] < min:
            min = arr[i]
        else:
            profit = arr[i] - min
            if profit > max:
                max = profit
            
    return max

#input
arr = list(map(int, input().split()))
print(max_profit(arr))
```
