Input: 1234
Output: 4321

Costly Item from a Bill

```python
a = int(input())
max = 0
cost_item = " "
for _ in range(a):
    entry = input().split()
    item = entry[0]+" "+entry[1]
    price = int(entry[-1])

    if price > max:
        max = price 
        cost_item = item

print(cost_item , max)
```

Find the next prime number greater than N

```python
a = int(input())

def is_prime(a):
    if a <=1:
        return False
    i = 2
    while i*i <= a:
        if a%i ==0:
            return False
        i+=1
    return True

num = a+1
while True:
    if is_prime(num):
        print(num)
        break
    num +=1
```

What is a Perfect Number?
A perfect number is a positive integer that is equal to the sum of its proper divisors (excluding itself).

```python


n = int(input())
count =0
for i in range(1,n):
    if n %i == 0:
        count +=i
print(count)
```

35. Modify array by divisibility

```python
arr = list(map(int,input().split()))
res =[]
for x in arr:
    if x%3==0 and x%5 ==0:
        res.append("threefive")
    elif x%3 ==0:
        res.append("three")
    elif x%5==0:
        res.append("five")
    else:
        res.append(x)

print(res)
```
