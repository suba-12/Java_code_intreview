Find Average and Sum of an Array
Problem Statement:Given an array of numbers, find the average and sum ofthe elements.
Example:
Input:
1 2 3 4 5
Output:
Sum = 15
Average = 3

```python
num = list(map(int, input().split()))
n = len(num)
sum = 0
for i in num:
    sum = sum+i

avg = sum // n
print('Sum = ' ,sum)
print('Average =' , avg)

```

Find all palindrome numbers in the range of A to B

```python
a = int(input())
b = int(input())

for i in range(a , b+1):
    if str(i) == str(i) [::-1]:
        print(i ,end =' ')
```
