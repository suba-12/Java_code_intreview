Total Number of Subarrays 

```python
arr= list(map(int,input().split()))
target= int(input())

def subarray(arr,target):
    n= len(arr)
    count=0
    for i in range(n):
        cur =0
        for j in range(i,n):
            cur =cur+arr[j]

            if cur == target:
                count+=1
    return count

result = subarray(arr,target)
print(result)

```

2. Rotate a Matrix by 90 Degrees Clockwise

```python
n = int(input())
arr=[]
for _ in range(n):
    arr.append(list(map(int , input().split())))


def roatate(n,arr):
    #transpose
    for i in range(n):
        for j in range(i,n):
            arr[i][j] , arr[j][i] = arr[j][i] , arr[i][j]
            #reverse
    for row in arr:
        row[:] = row[::-1] # use row.reverse()

roatate(n,arr)
print(arr)
```

```python
n = int(input())
elements = list(map(int,input().split()))
arr =[elements[i*n:(i+1)*n] for i in range(n)]

def rotate(n ,arr):
    for i in range(n):
        for j in range(i,n):
            arr[i][j] , arr[j][i] = arr[j][i] , arr[i][j]

    for row in arr:
        row.reverse()

rotate(n,arr)
print(arr)
```

3 .Rotate the matrix

```python
arr = list(map(int,input().split()))
k = int(input())

def rotate(arr,k):
    n = len(arr)
    for _ in range(k):
        last = arr[-1]
        for i in range(n-1 , 0 , -1):
            arr[i] = arr[i-1]
        arr[0] = last

rotate(arr,k)
print(arr)
```

Maximum Subarray Sum - Brute force

```python
arr = list(map(int,input().split()))
def subarray(arr):
    max = arr[0]
    cur = 0
    n = len(arr)
    for i in range(n):
        cur = 0
        for j in range(i,n):
            cur +=arr[j]
            if cur > max:
                max = cur

    return max
subarray(arr)
```

 5. Count of Elements Greater Than All Prior Elements

```python
arr = list(map(int,input().split()))
def prior(arr):
    count = 0
    max = float('-inf')
    for i in arr:
        if i > max:
            count +=1
            max = i

    return count
print(prior(arr))
```

Push Empty Packets (0) to the End

```python
arr = list(map(int,input().split()))
def push_zeros(arr):
    non_zeros =[x for x in arr if x != 0]
    zero = [0]*(len(arr)-len(non_zeros))
    return non_zeros +zero
print(push_zeros(arr))
```

Two Sum – Find Indexes of Two Elements Adding to Target

```python
arr = list(map(int,input().split()))
target = int(input())
n = len(arr)
def sum(arr,target):
    for i in range(n):
        for j in range(i+1,n):
            if arr[i]+arr[j] ==target:
                return [i, j]
        return []
print(sum(arr,target))
        
```

Pascal's Triangle

```python
n = int(input())
res = []
for i in range(n):
    row =[1]*(i+1)
    for j in range(1,i):
        row[j] = res[i-1][j-1]+res[i-1][j]
    res.append(row)

print(res)
```
